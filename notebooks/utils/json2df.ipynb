```python
from json import load
import pandas as pd
```

```python
path = '/Users/samedi/Desktop/tgW.json'
```

```python
j = load(open(path,'r'))[0]['events']
df = pd.DataFrame(data={
    k: list(map(lambda x: x[k], j)) for k in ["id","class","start","end","summative work time","percentage working"]
})
df.head()
```

```python
dfn = {'id': [], 'notes':[]}
for row in j:
    for note in row['notes']:
        dfn['id'].append(row['id'])
        dfn['notes'].append(note)
dfn = pd.DataFrame(dfn)
dfn.head()
```

```python

dfa = {'id': [], 'start':[], 'end':[]}
for row in j:
    for event in row['work events']:
        dfa['id'].append(row['id'])
        dfa['start'].append(event["start work"])
        dfa['end'].append(event["stop work"])
dfa = pd.DataFrame(dfa)
dfa.head()
```

```python
df.to_csv('tracks.csv')
dfn.to_csv('notes.csv')
dfa.to_csv('events.csv')
```
