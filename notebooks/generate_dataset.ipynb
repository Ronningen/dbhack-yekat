### Создание датасета формата yolo из нашей разметки

```python
path_viame_dataset = r"G:\Мой диск\5min\get"
# path_viame_dataset = r"C:\workspace\hakaton\hak2023_Digital_breakthrough_ekat\sources\val"
path_yolo_dataset = r"C:\workspace\hakaton\hak2023_Digital_breakthrough_ekat\dataset_yolo_2206\train"
```

```python
from training.dataset import ViameReader

accum_class = {'crane': ['high'], 'earth': ['unknown']} # high будем переводить в класс crane
name_list = ['earth', 'bobcat', 'crane', 'loader', 'truck'] # id будут выставлены согласно списку, и далее для новых классов
vr = ViameReader(path_viame_dataset, img_size_out=1280,
                 path_out_dir=path_yolo_dataset,
                 class_accumulation=accum_class,
                 class_names=name_list,
                 step_frame=15)
print("Все классы датасета:")
vr.get_all_class()
```

```python
import os

vr.start_processing(out_format='bbox_yolo')
print(vr.class_id2class_name)
print(list(vr.class_name2class_id), "; count =", vr.len_class)
with open(os.path.join(path_yolo_dataset, "class_name.txt"), "w") as f:
    print(vr.class_id2class_name, file=f)
    print(list(vr.class_name2class_id), "; count =", vr.len_class, file=f)
# vr.get_all_class()
```
